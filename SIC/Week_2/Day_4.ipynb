```python
import random

Int_1 = random.randint(0, 10)   # Inclusive     ;to return a integer number at the range 0 and 10.  
print(f"The output of randint(): {Int_1}")

list_numbers = [i**3 for i in range(1, 11)]
print(f"The original list is: {list_numbers}")

Int_2 = random.choice(list_numbers)     # this randomly chooses;    if done this on a for loop, it ofc does not keep track of what was previously added.

print(f"The output of choice(list): {Int_2}")

list_random = [random.choice(list_numbers) for _ in range(10)]    # obtain a list of random elements (with elements having a limit).
print(f"1) The list after choice(list) on for loop is: {list_random}")

list_random = random.shuffle(list_numbers)
print(f"2) The list after using shuffle(list): {list_random}")  # it shuffles a list whithout needing a for loop over choice.

list_random = random.sample(list_numbers, 7)    # obtain 'N' elements randomly from one list,  
print(f"The list after using sample() on for loop is: {list_random}")   # May use above shuffle fn along with slicing, to obtain the sample feature of custom elements.

# testing the functionality of randrange.
print()     # slip a line at terminal 
for _ in range(10):
    Int_3 = random.randrange(0, 5, 1)   # follows the 'range()' parameter's effects.    ; high argument is exclusive, 3rd argument is step (to select elements from a subset of given range, i.e. the step value)
    print(f"The output of randrange() is: {Int_3}")

# final fn for this session- uniform()  -> returns a real number from the given range.
print()
for _ in range(10):
    Int_4 = random.uniform(1, 10)
    print(f"The single real number obtained from  uniform() fn: {Int_4}")
```

```python
# Dice game

import random
import time
def roll_dice(n):
    dice = [] # start with empty list of dice
    # add random numbers between 1 to 6 to the list
    for i in range(n):
        dice.append(random.randint(1,6))
    return dice

def find_winner(cdice_list, udice_list):
    computer_total = sum(cdice_list)
    user_total = sum(udice_list)
    
    print('Computer total', computer_total)
    print('User total', user_total )
    
    if user_total > computer_total:
        print('User wins')
    elif user_total < computer_total:
        print('Computer wins')
    else:
        print('It is a tie!')
        
def roll_again(choices, dice_list):
    
    print('Rolling again ...')
    time.sleep(3)
    
    for i in range(len(choices)):
        if choices[i] == 'r':
            dice_list[i] = random.randint(1,6)
    
    time.sleep(3)
    

def computer_strategy1(n):
    """Strategy: Roll everything again."""
    print('Computer is thinking...')
    time.sleep(1.5)
    choices = ""
    for i in range(n):
        choices = choices + 'r'
    return choices

def computer_strategy2(n, computer_rolls):
    """Strategy: Roll again if the die value is less than 5."""
    print('Computer is thinking...')
    time.sleep(1.5)
    choices = ""
    for i in range(n):
        if computer_rolls[i] < 5:
            choices = choices + 'r'
        else:
            choices = choices + '-'
    return choices

# --- Main Program Area ---

# Step 1: Initialization
number_dice = int(input('Enter number of dice: '))
ready = input('Ready to start? Hit any key to continue ')

# Step 2: User turn
user_rolls = roll_dice(number_dice)
print('User first roll: ', user_rolls)

# Step 4: Get user choices
user_choices = input("Enter '-' to hold or 'r' to roll again: ").lower()

# Check length of user input
while len(user_choices) != number_dice:
    print(f'You must enter {number_dice} choices')
    user_choices = input("Enter '-' to hold or 'r' to roll again: ").lower()

# Step 5: Process user choices
roll_again(user_choices, user_rolls)
print('Player new Roll: ', user_rolls)

print('-' * 20)

# Computer's turn
print("Computer's turn")
computer_rolls = roll_dice(number_dice)
print('Computer first roll: ', computer_rolls)

# Step 6: Decide strategy
# We pass computer_rolls so the strategy can "see" the dice
computer_choices = computer_strategy2(number_dice, computer_rolls)
print('Computer Choice: ', computer_choices)

# Computer rolls again
roll_again(computer_choices, computer_rolls)
print('Computer new Roll: ', computer_rolls)

# Final Result
find_winner(computer_rolls, user_rolls)
```

```python
# You can create a series, which is a pseudo list and dict, can access a series via index, or if key is not int, access via a string.
    # create series by list
    # create series by py dict
    # create series by numpy library
    # create series by scalar value.
    
# A pandas series is a numpy backed array, however with indexes.
    # The native definition of a pandas series is the exact numpy array, then adding indexes to them.

# So, in conclusion- Pandas creates an array, that is as contigious block of memory. This method is called 'Vectorized layout'.
    # native python dictionary, creates a python object for every elemet in dictionary.
        # however pandas stores the 'type' of the whole Series once, and only the bit values are stored for all elements in list.

import pandas

list_ = [i*i for i in range(1, 11)]
pandas_Series = pandas.Series(list_)
print(pandas_Series)
 
temp = "abcdefghij"
dict_ = {}
j = 1
for i in temp:
    dict_[i] = j*j
    j += 1

pandas_Series = pandas.Series(dict_)
print(f"\nThe series in dict is: \n{pandas_Series}")
```

```python
import pandas

# Creating a sample Series to work with
# We include some duplicates and a None (NaN) to see how functions behave
s = pandas.Series([10, 20, 20, 30, 40, 50, None], name="SampleData")

# 1. s.count() - Returns the number of non-NA/null observations
print(f"1. Count (non-null): {s.count()}")

# 2. s.sum() - Returns the sum of the values
print(f"2. Sum: {s.sum()}")

# 3. s.mean() - Returns the average of the values
print(f"3. Mean (Average): {s.mean()}")

# 4. s.median() - Returns the middle value (50th percentile)
print(f"4. Median: {s.median()}")

# 5. s.min() & s.max() - Returns the smallest and largest values
print(f"5. Min: {s.min()} | Max: {s.max()}")

# 6. s.std() - Returns the standard deviation (measure of data spread)
print(f"6. Standard Deviation: {s.std():.2f}")

# 7. s.unique() - Returns an array of the unique items (removes duplicates)
print(f"7. Unique values: {s.unique()}")

# 8. s.nunique() - Returns the NUMBER of unique items (ignores NaNs)
print(f"8. Number of unique items: {s.nunique()}")

# 9. s.value_counts() - Returns how many times each unique value appears
print(f"\n9. Value Counts:\n{s.value_counts()}")

# 10. s.describe() - The "All-in-One": Returns a Series of several statistics
print(f"\n10. Full Describe Summary:\n{s.describe()}")
```
