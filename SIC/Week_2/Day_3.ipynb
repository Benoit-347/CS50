```python
# N queens final code:
# we will attempt by placing from 1 row to the next, So row index is in our solution space is a constant. Hence ignoring the N**2 by eliminating rows; as this will be symmetrical to the original.
def check_not_overlap(i, col):  # i is the row no. (which is also the queens placed count)
    for k in range(i):
        # overlap condition
        if (col[i]==col[k]  or  abs(col[i]-col[k]) == abs(i-k)):    # putting abs outside and removing the each abs is errornaeous.
            return False
    return True

# initially the i val is -1, then first iter i is incr to 0 and recurse is called.
def n_queens(i, col):
    
    # only body of this fn is this recur ender Base Cond.
    if check_not_overlap(i, col):
        if i == len(col) -1:
            print(f"\nFound a solution!:\n{col}")
        else:
            for j in range(len(col)):   # Until ALL QUEENs are placed (ALL ROWS)
                col[i+1] = j
                temp_filter = [i for i in col if i != -1]
                print(f"Traversing {col}, recur counter = {len(temp_filter)}")
                n_queens(i+1, col)
    else:
        print(f"Eliminated {col}")

n_queens(-1, [-1]*4)
```

```python
# TOPIC: Greedy algorithm
# We use this for selection among huge possibilities. This algo is the optimization in selection process (as was backtracking)
    # This selects by max first, giving priority to highest first, meeting all constraints.
    # 
# Coin selection min

def select_min_coins(list_coin_meta, amount):
    result = []
    index_coin_meta = 0
    while (amount > 0):
        count_coins = amount // list_coin_meta[index_coin_meta]     # this high level of optimization is possible only because of the division operator allowed in our CPU's (instead of multiple subtractions)
        if count_coins > 0:    # if list_coin_meta[index_coin_meta] < amount:
            result.append(count_coins)
            amount -= count_coins * list_coin_meta[index_coin_meta]
        else:
            result.append(0)
        index_coin_meta += 1
    
    return result

coins_meta_list = [100, 50, 20, 10, 5, 2, 1]
Change_list = select_min_coins(coins_meta_list, 120)
print(f"The change to be given are:\n")
j = 0
for i in Change_list:
    if i != 0:
        print(f"Coin- {coins_meta_list[j]} \t count: {i}")
    j += 1
```

```python
# Given a single room, tht is booked by multiple events, maximize the no of events booked.
# given: start time and end time of the booked event

def activty_selection(start, finish):
    result = []
    index_event = 0
    result.append(index_event)
    for j in range(1, len(start)):
        if finish[index_event] <= start[j]:
            result.append(j)
            index_event = j
    return result

list_of_events = [ [1, 2], [3, 4], [2, 5], [0, 6], [5, 6], [8, 9], [5, 9] ]
start_indexes = [i[0] for i in list_of_events]
end_indexes = [i[1] for i in list_of_events]

Answer = activty_selection(start_indexes, end_indexes)
print(f"The ID's of the activities selected are: {Answer};\t Count is: {len(Answer)}")
```

```python
# fibonnaci (iterative would be faster) using recursion, with DP
Cache = {0: 0, 1: 1}

def fibonacci_DP(n):
    if n in Cache:
        return Cache[n]
    else:
        Cache[n] = fibonacci_DP(n-1) + fibonacci_DP(n-2)
        return Cache[n]
    
import random
n = random.randint(1, 100)
Answer = fibonacci_DP(n)
print(f"The fibonnaci value at {n} position is: {Answer}")
```

```python
# suppose the cache has fixed limit. And the fn itself handles the cache. We then have to initialize the cache for non-computed values. By convention in py, lets stick with -1.

def fibonacci_DP_2(Cache, n):
    
    # end case
    if n <= 1:
        return 1
    
    # check cache
    if Cache[n] != -1:
        return Cache[n]
    
    result = fibonacci_DP_2(Cache, n-1) + fibonacci_DP_2(Cache, n-2)
    Cache[n] = result
    return result

import random
n = random.randint(1, 100)

Cache = [0, 1] + [-1]*(n-1)     # first fibbo is 1, excluding tht we have (n-1)
Answer = fibonacci_DP_2(Cache, n)

print(f"The fibonnaci value at {n} position is: {Answer}")
```

```python
# pascal's triangle:
def bin_coeff(n, k):
    if k == 0 or n ==k:
        return 1
    else:
        return bin_coeff(n-1, k-1) + bin_coeff(n-1, k)
    
import random
n = random.randint(1, 30)
k = random.randint(1, n)    # k is always equal to or less than n (randint() is inclusive)

Answer = bin_coeff(n, k)

print(f"The binary cofficient value at {n, k} position is: {Answer}")
```

```python
def bin_coeff_DP(n, k):
    B = [0]* (n+1)
    for i in range(n+1):
        for j in range(n, -1, -1):
            if j == 0 or j == i:
                B[j] = 1
            else:
                B[j] = B[j] + B[j-1]
    return B[k]

import random
n = random.randint(1, 100)
k = random.randint(1, n)

Answer = bin_coeff_DP(n, k)

print(f"The binary cofficient value at {n, k} position is: {Answer}")
```
