```python
# One of Easiest Sorting algo- Bubble Sort
# key method- Traverse from left of list, cmp two elements as you go. If next is lower thn cur, swap.
# Double Loop and perform key method at 2nd depth of loop.
# Result: The last element is sorted, N last elements are sorted for N iternation done.
# Very bad efficiency due to 2 list operations for n*(n/2) iterations
import random
list_1 = [76, 82, 66, 14, 28, 18, 96, 3, 59, 72]
"""
for i in range(10):
    list_1.append(random.randint(1, 100))
"""
N = len(list_1)

for i in range(0, N):
    for j in range(0, N-1):
        if  list_1[j] > list_1[j+1]:
            list_1[j], list_1[j+1] = list_1[j+1], list_1[j]
    
    print(f"At iteration {i+1} the input list now is {list_1}")
            
```

```python
# Selection Sort
# Also One of the easiet sorting algorithms in terms of finding min/max first, then inserting
# Every iteration gets a min/max at end, keeping a sorted section, And the following sections, does not insert bw the sorted section, 
    # however different from bubble sort that, swaping is not the main operation in every iteration, but finding the end element first.
def min(list_1):
    if not list_1:
        return False
    min_ = list_1[0]
    for i in list_1:
        if min_ > i:
            min_ = i
    return min_

input_list = [76, 82, 66, 14, 28, 18, 96, 3, 59, 72]
result = []
j = 1

# note: whn doing a iterable on for loop, if you modify the iterable, the for loop will also be affected
N = len(input_list)
for i in range(N):
    min_val = min(input_list)
    result.append(min_val)
    input_list.pop(input_list.index(min_val))
    print(f"The sorted list at iteration {j} is: {result}")    
    j+=1
```

```python
# Sorting without additional memory consumption:

input_list = [76, 82, 66, 14, 28, 18, 96, 3, 59, 72]
N = len(input_list)

# we implement a custom min fn, as we want to find 1st min, 2nd min... So every iteration, we find most min, until least min.
for i in range(N):
    min_ = input_list[i]
    for j in range(i+1, N):      # we need to skip the first part of list, as we will later keep fully sorted elements here.
        if min_ > input_list[j]:
            min_ = input_list[j]
    print(f"Found min now is: {min_}")
    min_index = input_list.index(min_)
    input_list[i], input_list[min_index] = min_, input_list[i]
    print(f"The input list at iteration {i+1} is now: {input_list}")
            
```

```python
# Insertion sort
input_list = [76, 82, 66, 14, 28, 18, 96, 3, 59, 72]
N = len(input_list)

# Every iteration, we dont find the sorted section completely. But we keep a list tht is currently sorted, with i elements; where future elements from input MAY insert into this sorted section.
result = []
len_result = 0
# first iter direct append will happen
for i in range(N):
    for j in range(len_result):     # the result list itself is is sorted state, so just traversing once finds insert pos. 
        # add new number either at between, if new element makes the list unsorted.
            # else if its never in betweem, put the element at end.
        if result[j] > input_list[i]:
            # starting logic for insertion
            result.append(result[j])    # basically a assign new, then shift to make space, then assign actual val
            for k in range(len_result-1, j-1, -1):
                result[k+1] = result[k]
            result[j] = input_list[i]
            # ending logic for insertion
            break
    else:
        result.append(input_list[i])
    len_result += 1
    print(f"At iteration {i+1} the insertion list is: {result}")
```

```python
# 2. Merge

def merge_by_double_traversal(list_1_sorted, list_2_sorted):
    result = []
    
    # will do a single iteration, hence we need a number that covers both lists completely hence taking greater of the two.
    N = len(list_1_sorted)
    M = len(list_2_sorted)
    
    index_1 = 0
    index_2 = 0
    for _ in range(N + M):      # loop until the smaller list is exhausted
        ele_1 = list_1_sorted[index_1]
        ele_2 = list_2_sorted[index_2]
        if ele_1 > ele_2:
            result.append(ele_2)
            index_2 += 1 
        else:
            result.append(ele_1)
            index_1 += 1
        if (index_1 >= N) or (index_2 >= M):
            break
    
    # Remaining elements
    
    while (index_1 < N):
        result.append(list_1_sorted[index_1])
        index_1 += 1
    # either above or below will pass while cond, not both. (If both thn our main for loop logic would be wrong haha)
    while (index_2 < N):
        result.append(list_2_sorted[index_2])
        index_2 += 1
        
    return result


list_1_sorted = [i**2 for i in range(1, 7)]
list_2_sorted = [i**3 for i in range(1, 7)]
print(merge_by_double_traversal(list_1_sorted, list_2_sorted))
```
