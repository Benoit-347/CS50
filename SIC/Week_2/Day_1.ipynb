```python
# revising class

# class creation
class temp:
    pass

# obj creation
obj_a = temp()
print(id(obj_a))

# class instance variables (it is apart from object var)
class temp_2:
    class_var_1 = "Data is from class_var_1"    # this is shared among all objects created from this class.

# adding instances to a class during definition
    # NOT possible in python for a class to create instances for an object without first creating the object.
    # hence the way is to use a constructor to create instance variables of a class.

# you cannot create variables without the user assigning it values.

# *WRONG* 
# class temp_2:
#   name_1
#   name_2

# *WRONG*
# class temp_2:
#   def __init__(self):
#       self.name_1
#       self.name_2
# var_a     *WRONG*

class temp_3:
    def __init__(self):
        self.var_1 = "data of var_1"
        self.var_2 = "data of var_2"

obj_b = temp_3()
print(obj_b.var_1)     
print(obj_b.var_2)

# also in py, a defined class is not fixed by intepretor. Hence any 'fixed' classes are programmer imposed. (Which is exactly what a constructor does) 
                        # In low-level terms—specifically regarding memory management (__dict__ is same) and how the CPU interacts with data. However the IDE can recogize the names easier AND,
                            # consistency by programmer, + arguments give error if programmer makes mistake. (Manual is messy).
                            
obj_b.name_3 = "data of name_3"
obj_b.name_4 = "data of name_4"

print(obj_b.name_3)
print(obj_b.name_4)

# __init__ means constructor
# __str__ means obj behavior changed to return string.
# these are called special methods, they are called automatically by the intepretor.

# variable names with '__varName' are used by convention in python, to hide these variables from user access. i.e. after object creation, so only the class methods can access them.

# __dict__ "It is a mapping object in python, that stores a oject's writable attributes."
# Every time you do a dot operation in python, this mapping object is what is actually reffered.
# However Python does have 2 __dicts__:     i) for classes- has 1 __dict__ for each class defined,    ii) objects- has 1 __dict__ for each object, so N objects of a class = N  __dict__'s
    # contains all instance variables and methods of the object/class.
    
# also operator overloading also done similar syntax as above
# __add__ "Overload '+' operator to our defined fn within our classs"
# __sub__  "        '-' "           "           "           "
# __mul__  "        '*' "           "           "           "
# __truediv__       '/' "           "           "           "
# __floordiv__      '//' "           "           "           "
# __mod__           '%' "           "           "           "
# __pow__           '^' "           "           "           "

# additional-
    # '+X' the prefix with only 1 '+'   ->  __pos__
    # '-x' the prefix with only 1 '-'   ->  __neg__
```
